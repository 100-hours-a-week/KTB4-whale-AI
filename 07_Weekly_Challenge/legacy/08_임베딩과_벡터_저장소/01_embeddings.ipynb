```python
# 구현체 생성
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key="YOUR_API_KEY",
)
```

```python
# 여러 문서를 한 번에 벡터로 변환
texts = [
    "어댑터즈는 스타트업코드에서 제공하는 개발 책 서빙 서비스입니다.",
    "어댑터즈는 개발자 교육 콘텐츠를 제공합니다.",
    "오늘 날씨가 정말 맑습니다.",
]

vectors = embeddings.embed_documents(texts)

print(f"문서 수: {len(vectors)}")
print(f"벡터 차원: {len(vectors[0])}")
print(f"\n첫 번째 벡터의 앞 5개 값: {vectors[0][:5]}")
```

```python
# 단일 질의를 벡터로 변환
query = "어댑터즈는 어떤 서비스인가요?"

query_vector = embeddings.embed_query(query)

print(f"벡터 차원: {len(query_vector)}")
print(f"앞 5개 값: {query_vector[:5]}")
```

```python
# 유사도 계산으로 의미 거리 확인
import numpy as np

def cosine(a, b):
    a, b = np.array(a), np.array(b)
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

# 3번에서 만든 vectors와 4번에서 만든 query_vector 사용
sim_0 = cosine(query_vector, vectors[0])  # 어댑터즈 = 개발 책 서빙 서비스
sim_1 = cosine(query_vector, vectors[1])  # 어댑터즈 = 개발자 교육 콘텐츠
sim_2 = cosine(query_vector, vectors[2])  # 오늘 날씨가 맑습니다

print(f"질의 vs 문서0 (어댑터즈 소개):  {sim_0:.4f}")
print(f"질의 vs 문서1 (어댑터즈 콘텐츠): {sim_1:.4f}")
print(f"질의 vs 문서2 (날씨):           {sim_2:.4f}")
```
